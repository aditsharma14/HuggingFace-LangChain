```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="conversational",
    max_new_tokens=200,
    temperature=0.3,
    top_p=0.95,
    huggingfacehub_api_token=os.getenv("hf_token")
)
chat = ChatHuggingFace(llm=llm)
```

```python
chat.invoke("What is Machine Learning?").content
```

```python
chat.invoke("What is generative AI ").content
```

```python
repoid="google/gemma-3-12b-it"
llm = HuggingFaceEndpoint(
    repo_id=repoid,
    task="conversational",
    max_new_tokens=200,
    temperature=0.3,
    top_p=0.95,
    huggingfacehub_api_token=os.getenv("hf_token")
)
chat = ChatHuggingFace(llm=llm)
```

```python
chat.invoke("What was the white revolution in India?")
```

```python
from langchain_core.prompts import PromptTemplate
template="""
Question:{question}
Answer:Lets think step by step.
"""
prompt=PromptTemplate(template=template,input_variables=["question"])
print(prompt)
```

```python
chain = prompt | chat
chain.invoke({"question": "What was the white revolution in India?"}).content
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

os.environ["HF_TOKEN"] = os.getenv("hf_token")  # authenticated downloads from the HF Hub

model_name = "BAAI/bge-small-en"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": True}
hf = HuggingFaceEmbeddings(
    model_name=model_name, model_kwargs=model_kwargs, encode_kwargs=encode_kwargs
)
```

```python
embedding = hf.embed_query("hi this is alex")
```
